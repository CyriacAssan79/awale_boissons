# Note d'adoption interne — reporting mensuel Awalé Boissons

*Pour Kômian : le community manager (CM) et le directeur de comptes (DC). Une page. Rédigée après avoir rejoué le pipeline de bout en bout ; les durées sont celles du Runbook (`docs/Runbook_Mensuel.ipynb`, §11).*

## 1. Qui fait quoi, dans quel ordre, en combien de temps

| # | Étape | Qui | Durée |
|---|---|---|---|
| 1 | Déposer le fichier du mois dans `data/raw/` (mêmes feuilles et colonnes que le mois précédent) | CM | 10 min |
| 2 | Lancer `python run_pipeline.py` | CM | 1 min de saisie |
| 3 | Attendre : chargement 3 s, transformation 21 s, tests 10 s, puis **classification IA ~45 min** (≈ 470 nouveaux commentaires). La machine travaille seule et sauvegarde toutes les ~2 minutes. | machine | ~46 min sans surveillance |
| 4 | Lire le bilan du lancement (`[OK]` ou `[ERREUR]`) et les contrôles qualité : jours manquants, montants WhatsApp, écarts export / plan média | CM, puis DC | 10 min |
| 5 | Ouvrir le dashboard (`streamlit run app/app.py`) et parcourir les 4 blocs | CM | 5 min |
| 6 | Valider l'interprétation, la recommandation et la note à Aïcha | DC | non mesuré : à chronométrer au premier cycle réel |

**Total ≈ 70 min, dont 25 min de présence active du CM** : au-dessus de l'heure visée par le brief, le temps venant presque entièrement de l'IA locale. Le premier lancement classe tout l'historique (2 831 commentaires, ~4 h 20) : à planifier hors cycle.

## 2. Formation de 45 minutes (CM et DC ensemble)

| Minutes | Contenu |
|---|---|
| 0 – 5 | Ce que le rapport dit et **ne dit pas** : une association dans le temps n'est pas une cause. Exemple : « prix » très négatif ne prouve pas qu'un canal en est responsable. |
| 5 – 15 | Démonstration du cycle complet sur des données d'essai. Le CM le refait seul, le DC observe. |
| 15 – 25 | Lire les 4 blocs et les **avertissements** : mois incomplet, montants sans valeur, écarts export / plan, commentaires « hors spam ». |
| 25 – 35 | Quand ça échoue : cinq messages réels et la réponse à chacun (feuille vide, colonne manquante, fichier de base ouvert dans un autre programme, test qui échoue sur un SKU inconnu, IA interrompue). |
| 35 – 42 | Exercice : un fichier avec un défaut volontaire (colonne renommée). Le CM trouve le message, corrige, relance ; le DC valide. |
| 42 – 45 | Qui appeler, quand ; remise de la fiche « si X alors Y » (`Runbook_Mensuel` §4 bis). |

## 3. Le risque d'abandon le plus probable au mois 3

**Le premier incident technique que le CM ne peut pas résoudre seul.** Le cycle demande un terminal, un environnement Python et 45 minutes de calcul ; le CM n'a jamais ouvert de terminal. Au mois 2, quelque chose change : une colonne renommée dans l'export, un magasin qui change d'identifiant (déjà arrivé au 1er mai), un aperçu de la base resté ouvert qui verrouille le fichier (arrivé pendant cet audit). Sans personne pour répondre ce jour-là, le réflexe est de revenir au tableur : deux jours, mais connu.

**Ce qui est déjà en place**

- Le chargement vérifie feuilles et colonnes **avant** d'écrire et nomme le problème ; il n'écrit rien en cas d'erreur.
- Les tests échouent avec un message explicite sur un SKU inconnu, un libellé de mois illisible ou un magasin qui vend sous deux identifiants le même jour, au lieu de produire un chiffre faux.
- L'IA reprend là où elle s'est arrêtée : un plantage ne fait plus perdre 45 minutes de calcul.
- Un mois de ventes incomplet est signalé dans le dashboard, sans intervention.

**Ce qui reste à faire (proposé, non réalisé)**

1. Un lanceur par double-clic qui affiche le résultat en français, pour que le CM n'ouvre pas de terminal.
2. Un référent technique nommé par le DC, joignable 30 minutes les jours 1 à 3 de chaque cycle, pendant les 3 premiers mois.
3. Un indicateur suivi chaque mois : « cycle lancé par le CM sans aide » (oui / non).

**Signal d'alerte :** à 30 jours, le CM n'a pas lancé un cycle seul ; ou deux incidents de suite ne sont résolus que par un ingénieur. Il faut alors revoir la conception, pas former davantage.
