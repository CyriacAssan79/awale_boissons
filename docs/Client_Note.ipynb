# Note de décision marketing

**Awalé Boissons — Janvier à juin 2026**


## 1. Synthèse exécutive

L'analyse des données disponibles documente les dépenses marketing, les ventes, les écarts entre plan média et dépenses observées, ainsi que les retours clients issus des réseaux sociaux.

- **Dépense campagne observée (janv.–juin) :** 15,84 M FCFA
- **dont Meta :** 6,95 M FCFA
- **dont TikTok :** 4,21 M FCFA

Le CA net observé varie fortement selon les mois. **Avril doit être interprété avec prudence** : 14 jours de données de ventes sont manquants (13 au 26 avril). Les jours sans données ne sont pas assimilés à des jours à zéro vente.

## 3. Customer Voice

L'inférence porte sur 2 831 commentaires Facebook, Instagram et TikTok. Le benchmark humain de 50 commentaires a servi à évaluer le classifieur ; il ne constitue pas un jeu d'entraînement. Sur le corpus complet, le modèle local a été utilisé pour les 2 831 commentaires.

| Signal | Volume / lecture |
|---|---|
| Commentaires | 2 831 |
| Positifs / négatifs | 1 186 / 735 |
| Spam détecté | 513 |
| Thèmes dominants | Goût 680 · Prix 376 · Promotion 313 · Disponibilité 170 |
| Prix | 347 mentions négatives sur 376 mentions du prix |
| Disponibilité | 151 mentions négatives sur 170 mentions |
| Goût | 578 mentions positives contre 14 négatives |

> **Lecture :** le prix et la disponibilité sont les principaux signaux négatifs du corpus. Le goût est très majoritairement positif. Ces observations décrivent la voix des clients ; elles ne démontrent pas un effet causal sur les ventes ou l'efficacité d'un canal publicitaire.

## 6. Limites et gouvernance

Les données permettent une lecture descriptive, mais pas une attribution causale directe des ventes aux canaux. Les écarts entre l'export campagne et le plan média doivent être réconciliés avant toute extrapolation. Les jours de ventes manquants restent manquants. Les classifications IA sont évaluées sur un benchmark humain séparé et doivent être suivies dans le temps.

**Principe de décision :**

```
Faits observés → signaux → limites → hypothèses → test
```

*L'objectif des prochains 90 jours est autant d'apprendre et de mieux mesurer que d'exécuter le budget.*
