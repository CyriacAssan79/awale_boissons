# Note de décision marketing

**Awalé Boissons — Janvier à juin 2026**


## 1. Recommandation

**Répartir les 15 M FCFA comme un budget de test sur 90 jours** (détail en 4), en gardant chaque canal financé mais en le conditionnant à une mesure. Les données ne permettent pas de dire quel canal fait vendre : elles servent à décider où mieux mesurer, pas où miser.

- **Dépense campagne observée (janv.–juin) :** 15,84 M FCFA, dont Meta 6,95 M et TikTok 4,21 M.
- **Avril est à lire avec prudence** : 14 jours de ventes manquent (13 au 26 avril) ; ils ne sont pas comptés comme des jours à zéro vente.

## 3. Customer Voice

2 831 commentaires (Facebook, Instagram, TikTok), classés par un modèle local évalué sur 50 commentaires annotés à la main. Sauf mention contraire, les comptes excluent les 513 commentaires détectés comme spam.

| Signal | Lecture |

Le prix et la disponibilité sont les principaux signaux négatifs, le goût est très majoritairement positif. Cela décrit la voix des clients ; cela ne démontre ni un effet sur les ventes ni l'efficacité d'un canal.

## 4. Proposition de test — 15 M FCFA, 90 jours

Budget de test pensé pour poursuivre l'activité tout en améliorant la mesure : ni classement des canaux, ni ROI. Calcul : socle de 1 M FCFA par canal pour l'instrumentation, puis le reliquat au prorata de la dépense observée, pondérée par la qualité de mesure ; recalculé à chaque cycle.

| Canal | Budget (FCFA) | Part | Condition de mesure |
|---|---|---|---|
| Meta | 5 550 000 | 37,0 % | Impressions, clics et conversion mesurable |
| TikTok | 3 100 000 | 20,7 % | Impressions, clics et conversion mesurable |
| Radio | 2 000 000 | 13,3 % | Code, numéro ou suivi dédié |
| Google | 1 900 000 | 12,7 % | Clics et conversion mesurable |
| Influenceurs | 1 450 000 | 9,7 % | Lien, code ou suivi dédié |
| Activation terrain | 1 000 000 | 6,7 % | Réconcilier facturation et dépenses avant extrapolation |
| **Total** | **15 000 000** | **100 %** | |

Le diagnostic est solide sur les dépenses digitales et les ventes, fragile sur la radio et l'activation terrain ; **rien ne prouve** aujourd'hui l'efficacité d'un canal.

## 6. Faut-il arrêter la radio ?

**Non, pas sur la base de ces données. Et pas non plus l'augmenter.**

La radio est le deuxième poste facturé (4,60 M FCFA, 26 % du facturé), mais l'export n'en enregistre que 2,00 M. Elle n'a ni impressions ni clics, et rien ne relie une diffusion à une commande. Ces données ne montrent donc ni qu'elle fonctionne, ni qu'elle ne fonctionne pas.

Nous proposons de **conserver 2,0 M FCFA (13,3 %) et d'y ajouter un numéro ou un code dédié**, afin que la question puisse être tranchée dans 90 jours. Si aucune commande ne porte ce code alors que le canal a été financé au niveau du test, la radio restera non mesurable : la couper ou la maintenir devra alors se décider sur d'autres critères que les données, et nous le dirons.

## 7. Ce que nous mesurerons dans 90 jours

| Mesure | Point de départ (janvier–juin 2026) | Ce que cela nous apprendra |
|---|---|---|
| Commandes WhatsApp portant un code de canal | **0** : aucun code n'existe aujourd'hui | Un canal avec au moins une commande codée devient mesurable ; sinon il reste non mesurable. |
| CA net par jour observé (points de vente) | 537 000 FCFA en moyenne sur les mois complets ; 402 000 en mai, 471 000 en juin | Semaines avec et sans campagne : un signal dans le temps, pas une preuve. |
| Commandes WhatsApp livrées par mois | 87, 100, 105, 71, 115, 146 | Tendance de la demande. |
| Part de commentaires négatifs sur le prix et sur la disponibilité (hors spam) | Prix : 92 % (347 sur 376) ; disponibilité : 89 % (151 sur 170) | Elles doivent baisser si prix ou stock sont traités. |
| Qualité des données | 14 jours de ventes manquants ; 96 commandes livrées sur 624 sans montant exploitable ; écart export / facturé : radio −2,60 M, activation terrain −2,41 M | Zéro jour manquant, écarts réconciliés, unité des montants WhatsApp confirmée. |

**Nous aurons eu raison** si, à 90 jours, au moins un des canaux aujourd'hui non mesurés est devenu mesurable et si les écarts de données sont réconciliés. **Nous aurons eu tort** si le budget a été dépensé sans qu'aucune de ces mesures ne progresse. Les seuils qui justifieraient de réduire ou d'augmenter un canal se fixent avec Aïcha **avant** le lancement.

## 8. Limites

Lecture descriptive uniquement : aucune attribution d'une vente à un canal. Les écarts entre l'export campagne et le plan média sont à réconcilier avant toute extrapolation, et les jours de ventes manquants restent manquants. Faits observés → signaux → limites → hypothèses → test : l'objectif des 90 prochains jours est autant d'apprendre et de mieux mesurer que de dépenser.
